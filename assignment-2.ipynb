```python
import numpy as np

marks = np.array([82, 67, 91, 74, 88, 56, 79, 95, 63, 71])

print("Marks:", marks)
print("Total marks:", np.sum(marks))
print("Average marks:", np.mean(marks))
print("Highest marks:", np.max(marks))
print("Lowest marks:", np.min(marks))
print("Marks greater than 75:", marks[marks > 75])
```

```python
import numpy as np

temperatures = np.array([28, 31, 29, 33, 35, 30, 32])
days = np.arange(1, 8)
updated_temperatures = temperatures + 2

print("Temperatures:", temperatures)
print("Average temperature:", np.mean(temperatures))
print("Highest temperature:", np.max(temperatures))
print("Lowest temperature:", np.min(temperatures))
print("Days above 30°C:", days[temperatures > 30])
print("Updated temperatures:", updated_temperatures)
```

```python
import pandas as pd

student_data = {
    "Name": ["Aarav", "Diya", "Rohan", "Ananya", "Vikram", "Meera", "Kabir", "Isha"],
    "Department": ["CSE", "ECE", "IT", "CSE", "ME", "ECE", "IT", "CSE"],
    "Marks": [85, 72, 91, 68, 77, 88, 54, 79],
    "Attendance": [92, 78, 85, 74, 81, 95, 69, 83]
}

students = pd.DataFrame(student_data)

print("First 5 students:")
print(students.head())
print("\nAverage marks:", students["Marks"].mean())
print("\nStudents who scored more than 75:")
print(students[students["Marks"] > 75])
print("\nStudents whose attendance is below 80%:")
print(students[students["Attendance"] < 80])
print("\nStudents sorted by marks:")
print(students.sort_values("Marks"))
```

```python
import pandas as pd

product_data = {
    "Product Name": ["Laptop", "Smartphone", "Headphones", "Keyboard", "Monitor", "Mouse"],
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Electronics", "Accessories"],
    "Price": [75000, 45000, 2500, 1800, 22000, 1200],
    "Quantity Sold": [18, 42, 75, 60, 25, 90]
}

products = pd.DataFrame(product_data)
products["Total Sales"] = products["Price"] * products["Quantity Sold"]

print("Product sales:")
print(products)
print("\nProduct with the highest sales:")
print(products.loc[products["Total Sales"].idxmax()])
print("\nAverage product price:", products["Price"].mean())
print("\nProducts with quantity sold greater than 50:")
print(products[products["Quantity Sold"] > 50])
print("\nProducts sorted by total sales:")
print(products.sort_values("Total Sales", ascending=False))
```

```python
import matplotlib.pyplot as plt

months = ["January", "February", "March", "April", "May", "June"]
sales = [12000, 15000, 13500, 18000, 21000, 19500]

plt.figure(figsize=(10, 5))
plt.plot(months, sales, marker="o", color="steelblue", label="Monthly Sales")
plt.title("Monthly Shop Sales")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))
plt.bar(months, sales, color="seagreen", label="Monthly Sales")
plt.title("Sales Comparison Across Months")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Aarav", "Diya", "Rohan", "Ananya", "Vikram", "Meera", "Kabir", "Isha", "Neha", "Arjun"]
marks = [85, 72, 91, 68, 77, 88, 35, 79, 56, 42]

plt.figure(figsize=(10, 5))
plt.bar(students, marks, color="cornflowerblue", label="Python Marks")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.legend()
plt.show()

categories = [
    "Excellent" if mark >= 80 else
    "Good" if mark >= 60 else
    "Average" if mark >= 40 else
    "Needs Improvement"
    for mark in marks
]
category_counts = {
    category: categories.count(category)
    for category in ["Excellent", "Good", "Average", "Needs Improvement"]
}

plt.figure(figsize=(8, 8))
plt.pie(category_counts.values(), labels=category_counts.keys(), autopct="%1.1f%%", startangle=90)
plt.title("Student Performance Distribution")
plt.show()
```

```python
"""
DSA Assignment Solutions
Contains solutions for Easy, Intermediate, and Difficult array problems.
"""

from collections import Counter
```

```python
# ==========================================
# EASY
# ==========================================

# 1. Reverse an Array
def reverse_array(arr):
    left = 0
    right = len(arr) - 1
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr
```

```python
# 2. Count Even, Odd and Zero
def count_even_odd_zero(arr):
    even = odd = zero = 0
    for num in arr:
        if num == 0:
            zero += 1
        elif num % 2 == 0:
            even += 1
        else:
            odd += 1
    print(f"Even: {even}\nOdd: {odd}\nZero: {zero}")
```

```python
# 3. Sum of Positive and Negative Numbers
def sum_pos_neg(arr):
    pos_sum = sum(x for x in arr if x > 0)
    neg_sum = sum(x for x in arr if x < 0)
    print(f"Positive Sum: {pos_sum}\nNegative Sum: {neg_sum}")
```

```python
# ==========================================
# INTERMEDIATE
# ==========================================

# 4. Remove Duplicate Elements
def remove_duplicates(arr):
    seen = set()
    result = []
    for num in arr:
        if num not in seen:
            seen.add(num)
            result.append(num)
    return result
```

```python
# 5. Find the Missing Number
def find_missing(arr):
    n = len(arr) + 1
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(arr)
    return expected_sum - actual_sum
```

```python
# 6. Rotate an Array
def rotate_array(arr, k):
    n = len(arr)
    if n == 0:
        return arr
    k = k % n
    return arr[-k:] + arr[:-k]
```

```python
# ==========================================
# DIFFICULT
# ==========================================

# 7. Find the Most Frequent Element
def most_frequent(arr):
    freq = {}
    for num in arr:
        freq[num] = freq.get(num, 0) + 1
        
    max_count = 0
    most_freq_elem = None
    
    for num, count in freq.items():
        if count > max_count:
            max_count = count
            most_freq_elem = num
            
    print(f"Most Frequent Element: {most_freq_elem}\nFrequency: {max_count}")
```

```python
# 8. Maximum Subarray Sum
def max_subarray(arr):
    max_sum = float('-inf')
    current_sum = 0
    start = end = s = 0
    
    for i, num in enumerate(arr):
        current_sum += num
        
        if current_sum > max_sum:
            max_sum = current_sum
            start = s
            end = i
            
        if current_sum < 0:
            current_sum = 0
            s = i + 1
            
    print(f"Maximum Subarray Sum: {max_sum}")
    print(f"Subarray:\n{arr[start:end+1]}")
```

```python
# 9. Check Whether Two Arrays are Equal
def check_arrays_equal(arr1, arr2):
    if Counter(arr1) == Counter(arr2):
        print("Arrays are Equal")
    else:
        print("Arrays are Not Equal")
```

```python
# ==========================================
# DRIVER CODE (Execution block)
# ==========================================
if __name__ == "__main__":
    print("--- 1. Reverse an Array ---")
    print(reverse_array([10, 20, 30, 40, 50]))
    print()

    print("--- 2. Count Even, Odd and Zero ---")
    count_even_odd_zero([10, 5, 0, 7, 8, 0, 13, 4])
    print()

    print("--- 3. Sum of Positive and Negative Numbers ---")
    sum_pos_neg([10, -5, 20, -8, 15, -2])
    print()

    print("--- 4. Remove Duplicate Elements ---")
    print(remove_duplicates([10, 20, 10, 30, 20, 40, 30]))
    print()

    print("--- 5. Find the Missing Number ---")
    print(f"Missing Number: {find_missing([1, 2, 3, 5, 6, 7])}")
    print()

    print("--- 6. Rotate an Array ---")
    print(rotate_array([1, 2, 3, 4, 5], 2))
    print()

    print("--- 7. Find the Most Frequent Element ---")
    most_frequent([2, 5, 2, 8, 5, 2, 3, 5, 2])
    print()

    print("--- 8. Maximum Subarray Sum ---")
    max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4])
    print()

    print("--- 9. Check Whether Two Arrays are Equal ---")
    check_arrays_equal([1, 2, 2, 3, 4], [4, 2, 1, 2, 3])
    print()
```
